```python
import warnings
```

```python
# What: this notebook is a thin driver that saves each timescale to
# its own .nc file (see quenchpath.io.save_tau_adv()/save_tau_chem()).
# Why: numerical/I-O logic lives in quenchpath.stages.stage_07_taus.
from quenchpath.config import GridResolution, RunConfig
from quenchpath.stages.stage_07_taus import (
    calc_and_save_adv_ls,
    calc_and_save_adv_qn,
    calc_and_save_chem_timescale,
)
```

```python
warnings.filterwarnings("ignore", module="iris")
```

```python
# What: choose experiment/resolution. tau_adv_ls is computed either
# way; tau_adv_qn only if exp="equ"; tau_chem_rls/tau_chem_net only
# if exp="kin".
# Why: tau_adv_qn needs that run's own species mole fraction field (see
# calc_and_save_adv_qn()'s docstring for more); tau_chem_rls/net need a
# completed merge-paths run at grid_resolution.
planet = "hd189733b"
exp = "kin"  # kin or equ
mdh = "0"
tK = "1"
grid_resolution = GridResolution.FULL
cfg = RunConfig(planet=planet, exp=exp, mdh=mdh, tK=tK)
```

```python
# Compute and save whichever timescales are valid for cfg.exp - see the
# cell above. Matches quenchpath.cli's own calc-taus branching.
tau_adv_ls = calc_and_save_adv_ls(cfg)
print(f"tau_adv_ls_dom: {tau_adv_ls['tau_adv_ls_dom'].data.min():.2e}-"
      f"{tau_adv_ls['tau_adv_ls_dom'].data.max():.2e} s")

if cfg.exp == "equ":
    tau_adv_qn = calc_and_save_adv_qn(cfg)
    print(f"tau_adv_qn: {tau_adv_qn['tau_adv_qn'].data.min():.2e}-"
          f"{tau_adv_qn['tau_adv_qn'].data.max():.2e} s")

if cfg.exp == "kin":
    tau_ch4_co_rls = calc_and_save_chem_timescale(cfg, grid_resolution, "rls")
    tau_ch4_co_net = calc_and_save_chem_timescale(cfg, grid_resolution, "net")
    print(f"tau_ch4_co_rls: {tau_ch4_co_rls.data.min():.2e}-{tau_ch4_co_rls.data.max():.2e} s")
    print(f"tau_ch4_co_net: {tau_ch4_co_net.data.min():.2e}-{tau_ch4_co_net.data.max():.2e} s")
```
