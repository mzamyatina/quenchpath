```python
# What: this notebook is a thin driver. See
# quenchpath.stages.stage_05_paths_parallel for a multiprocessing
# counterpart (recommended at FULL resolution).
# Why: numerical/I-O logic lives in
# quenchpath.stages.stage_05_paths_serial (I/O driver) and
# quenchpath.graph.shortest_path (pure, no iris/grid dependency).
from quenchpath.config import GridResolution, RunConfig, SpeciesConversion
from quenchpath.stages.stage_05_paths_serial import calc_paths_serial
```

```python
warnings.filterwarnings("ignore", module="iris")
```

```python
# What: choose experiment and grid resolution. GridResolution.SAMPLE is
# a coarse grid (~4896 locations/planet, ~30s); GridResolution.FULL is
# every native UM grid point (~855,360 locations/planet, ~20-25 min
# serially - use quenchpath.stages.stage_05_paths_parallel/
# `calc-paths-parallel` instead for a few minutes on a many-core
# machine).
# Why: SAMPLE/FULL write to separate output directories (see
# quenchpath.io.paths_dir()), so they can never collide.
planet = "hd189733b"
exp = "kin"
mdh = "0"
tK = "1"
cfg = RunConfig(planet=planet, exp=exp, mdh=mdh, tK=tK)
grid_resolution = GridResolution.SAMPLE
conversion = SpeciesConversion.CH4_CO  # or SpeciesConversion.CO_CH4
```

```python
# What: find and save the requested species conversion's shortest path
# for every location in the chosen grid.
# Why: equivalent to the original script's location loop, now inside
# calc_paths_serial().
results = list(calc_paths_serial(cfg, grid_resolution, conversion=conversion))
print(f"{len(results)} locations processed")
```
